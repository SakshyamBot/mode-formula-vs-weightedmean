```python
#absolute randomness
import random
err1 = 0
err2 = 0
#simulation model for mode
for i in range(1000):
  a = [random.randint(0,100) for i in range (0,1000)]
  #pegionhole principal guarentees atleast one number appears atleast 10 times
  freq = [len([a[j] for j in range(0,1000) if (i-10)<=a[j]<i]) for i in range(10,110,10)]
  #formula approach
  x = max(freq)
  ind = freq.index(x)
  y = freq[ind-1]
  z = freq[ind+1] if ind<len(freq)-1 else 0
  frequency = (ind)*10 + ((x-y)/(2*x - y-z))*10
  #my approach
  frequency_myway = ((ind*10 + 5)*x + (ind*10-5)*y + (ind*10+15)*z)/(x+y+z)
  #actual mode
  actual_mode = max(set(a), key=a.count)
  err1 += abs(frequency-actual_mode)
  err2 += abs(frequency_myway-actual_mode)
print(f"the errors by formula is {err1/1000} and error by my way is {err2/1000}")

```

```python
import random

err1 = 0
err2 = 0

for simulation in range(1000):
    mode = random.randint(0,100)
    a = [int(random.gauss(mode, 15)) for _ in range(1000)]
    a = [x for x in a if 0<=x<=100]
    freq = []
    for i in range(0, 100, 10):
        count = len([val for val in a if (i <= val < i + 10) or (i == 90 and val == 100)])
        freq.append(count)

    x = max(freq)
    ind = freq.index(x)

    y = freq[ind-1] if ind > 0 else 0
    z = freq[ind+1] if ind < len(freq)-1 else 0

    delta1 = x - y
    delta2 = x - z
    if (delta1 + delta2) == 0:
        frequency = ind * 10 + 5
    else:
        frequency = (ind * 10) + (delta1 / (delta1 + delta2)) * 10

    bin_center = ind * 10 + 5
    frequency_myway = ((bin_center * x) + ((bin_center - 10) * y) + ((bin_center + 10) * z)) / (x + y + z)


    err1 += abs(frequency - mode)
    err2 += abs(frequency_myway - mode)

print(f"The error by formula is {err1/1000:.2f}")
print(f"The error by my way is {err2/1000:.2f}")
```
